```python
X = df[audio_features]

print("Shape of clustering data:", X.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Scaled data shape:", X_scaled.shape)
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), inertia, marker='o')

plt.title('Elbow Method for Optimal Number of Clusters')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.xticks(range(2, 11))
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels,
        sample_size=5000,
        random_state=42
    )

    silhouette_scores.append(score)
    print("K =", k, "Silhouette Score =", round(score, 4))
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    silhouette_scores,
    marker='o'
)

plt.title('Silhouette Score for Different Numbers of Clusters')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.xticks(range(2, 11))
plt.grid(True)
plt.show()
```

```python
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)

df['cluster'] = kmeans.fit_predict(X_scaled)

print(df['cluster'].value_counts())
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def recommend_songs(song_name, n=10):

    matches = df[
        df['track_name'].str.lower() == song_name.lower()
    ]

    if len(matches) == 0:
        print("Song not found.")
        return

    song_index = matches.index[0]

    song_features = X_scaled[song_index].reshape(1, -1)

    similarity = cosine_similarity(
        song_features,
        X_scaled
    )[0]

    recommendations = df[
        ['track_name', 'track_artist',
         'playlist_genre', 'playlist_subgenre']
    ].copy()

    recommendations['similarity'] = similarity

    # Remove the input song and all duplicate copies
    input_artist = df.loc[song_index, 'track_artist']

    recommendations = recommendations[
        ~(
            (recommendations['track_name'].str.lower() == song_name.lower()) &
            (recommendations['track_artist'] == input_artist)
        )
    ]

    # Remove duplicate song + artist combinations
    recommendations = recommendations.drop_duplicates(
        subset=['track_name', 'track_artist']
    )

    recommendations = recommendations.sort_values(
        'similarity',
        ascending=False
    )

    return recommendations.head(n)
```

```python
def get_recommendations(song_name):

    if not song_name:
        return "Please enter a song name."

    recommendations = recommend_songs(song_name, n=10)

    if recommendations is None:
        return "Song not found. Please enter a song name from the dataset."

    result = "Recommended Songs:\n\n"

    for i, row in recommendations.reset_index(drop=True).iterrows():
        result += (
            f"{i+1}. {row['track_name']} — "
            f"{row['track_artist']} "
            f"(Similarity: {row['similarity']:.3f})\n"
        )

    return result


demo = gr.Interface(
    fn=get_recommendations,
    inputs=gr.Textbox(
        label="Enter Song Name",
        placeholder="Example: As If It's Your Last"
    ),
    outputs=gr.Textbox(
        label="Recommended Songs"
    ),
    title="🎵 Spotify Song Recommendation System",
    description="Enter a song name to get 10 similar songs based on audio features."
)

demo.launch()
```
